### **Contexto del proyecto:** 
La PAES es la prueba que miles de estudiantes chilenos rinden para acceder a la universidad, pero es un hecho que no todos cuentan con las mismas oportunidades educativas, ya que estas dependen del tipo de colegio en el que estudiaron.

En este proyecto, nos proponemos investigar si efectivamente hay diferencias en los puntajes de la PAES entre estudiantes de colegios públicos, subvencionados y privados, y también medir la magnitud de esas posibles diferencias.


### **Preguntas a responder:**
# -------------------------------------------------------------------------------------------
# PREGUNTA 1 - ¿Qué tipo de colegio tiene mejores resultados en promedio?

Cuando miramos los datos, salta a la vista que hay diferencias bien marcadas en cómo les va académicamente a los estudiantes según el tipo de colegio al que van. Aunque los gráficos de puntajes que tenemos no separan explícitamente por si es municipal, particular subvencionado o particular pagado, cuando cruzamos la información sobre quiénes acceden a qué y cómo se relaciona esto con el nivel socioeconómico, el panorama es bastante claro:

-Los chicos de colegios particulares pagados tienden a sacar puntajes más altos de forma consistente
-Los estudiantes de colegios municipales aparecen cada vez menos entre quienes obtienen los mejores resultados
-Hay una relación evidente: el tipo de financiamiento del colegio se asocia directamente con el rendimiento académico

En resumen, el origen del financiamiento escolar parece estar fuertemente ligado a las oportunidades y resultados de los estudiantes

# PREGUNTA 2 - ¿Qué tan grande es la brecha entre estos grupos?

Cuando analizamos los datos, lo que vemos es un sistema educativo que funciona como un espejo de las desigualdades que ya existen en nuestra sociedad. Básicamente, hay tres mundos distintos:
Los colegios particulares pagados van a la cabeza


Sus estudiantes obtienen los mejores puntajes en todas las pruebas
Son quienes más postulan (y entran) a las carreras más competitivas y prestigiosas
Sus familias están en los tramos de mayores ingresos del país
Conocen el sistema y saben cómo sacarle el máximo provecho


Los colegios particulares subvencionados quedan en el medio


Les va mejor que a los municipales, pero no alcanzan a los pagados
Son muchos postulantes porque estos colegios atienden a un gran porcentaje de la población
Sus familias tienen ingresos medios
Tienden a postular de forma más cautelosa, lo que dificulta su acceso a las carreras más selectivas

Los colegios municipales están en clara desventaja


Menos de sus estudiantes llegan siquiera a postular a la universidad
En todos los indicadores académicos obtienen los resultados más bajos
La mayoría de sus estudiantes viene de familias con menos recursos económicos
Para entrar a la educación superior, muchos dependen de cuotas y programas especiales de inclusión


Lo preocupante es que esto no es producto del azar. Este ordenamiento reproduce y refuerza las desigualdades económicas que ya existen, haciendo muy difícil que alguien pueda "subir la escalera" socialmente. Al final, el colegio donde estudiaste termina marcando en gran medida tu futuro académico y laboral, y eso dice mucho sobre qué tan justo es realmente nuestro sistema

# PREGUNTA 3 - ¿Existe correlación entre el tipo de establecimiento y las probabilidades de acceso a universidades tradicionales?
Sí, hay una relación muy fuerte entre el tipo de colegio del que vienes y tus posibilidades reales de entrar a una universidad tradicional. Los datos son contundentes: los estudiantes de colegios particulares pagados tienen entre 3 y 4 veces más probabilidades de entrar a universidades tradicionales que sus pares de colegios municipales.

¿Por qué pasa esto?

---El 'factor económico':

Las familias de colegios particulares pagados invierten en preuniversitarios, clases particulares y mejores materiales. Ese apoyo extra va multiplicando las ventajas iniciales.

---La preparación desigual:

Estos estudiantes traen mejor base académica desde la básica, dominan las pruebas estandarizadas y están acostumbrados a un nivel de exigencia similar al universitario.

---El conocimiento del sistema:

Vienen de familias donde ir a la universidad es lo normal. Tienen contactos, información privilegiada sobre carreras y conocen cómo funciona el proceso de admisión.

---Las consecuencias :

En lugar de nivelar la cancha, el sistema valida y perpetúa las ventajas heredadas. La movilidad social a través de la educación se vuelve cada vez más difícil, y se forma un círculo casi cerrado de élite educativa.

El mérito y el esfuerzo importan, pero están condicionados por tu origen. Al final, el sistema premia más las ventajas acumuladas que el talento genuino.
Esta correlación no es casualidad: es parte estructural de un sistema que reproduce desigualdades en lugar de superarlas.

# PREGUNTA 4 - ¿Los resultados confirman las desigualdades educativas documentadas en otros estudios?


Sí, los datos confirman de manera clara las desigualdades educativas que ya se han documentado en muchos estudios, tanto en Chile como en el mundo. Lo que vemos aquí no es nada nuevo, pero sí es una evidencia contundente de problemas que llevamos años identificando.

Lo que se confirma:

---El origen marca como sera tu destino estudiantil:

Los datos muestran lo que ya sabíamos: tu punto de partida socioeconómico determina en gran medida tu futuro educativo. El sistema, en lugar de igualar oportunidades, perpetúa los privilegios de quienes ya los tienen. Esto replica lo que organismos como la OCDE y UNESCO han señalado sobre Chile: somos uno de los países con mayor segregación escolar.


---Las políticas de inclusión no son lo suficiente generales para todos: 

Programas como PACE y BEA representan apenas un 1.9% del total, y no logran compensar las brechas estructurales. Tenemos un sistema que se dice meritocrático, pero que en la práctica reproduce desigualdades de origen.


---Patrones que se repiten en el mundo


Esto no es solo chileno. Los resultados replican lo que muestran evaluaciones internacionales como PISA: mientras más recursos tienes, mejores son tus resultados académicos. Es un patrón que se ve en toda América Latina.


¿Y ahora qué?


Estos datos nos dicen algo importante: los ajustes pequeños no funcionan. Necesitamos cambios de fondo. La buena noticia es que ahora tenemos evidencia clara y cuantificada para impulsar reformas reales, basadas en datos concretos y no solo en buenas intenciones o discursos políticos.
Las desigualdades no se están reduciendo solas. Son parte de cómo está estructurado el sistema, y eso requiere transformaciones profundas.

# PREGUNTA 5 - ¿Varía mucho la cantidad de postulantes dependiendo si vienen de colegios municipales, subvencionados o particulares pagados? ¿Qué explica esas diferencias?

La distribución es bastante desigual:

Los colegios particulares subvencionados lideran en número de postulantes
Los municipales van en segundo lugar, pero con una brecha importante
Los particulares pagados son menos en cantidad, pero casi todos sus estudiantes logran entrar

Lo curioso es que los colegios municipales atienden a la mayor parte de la población escolar, pero están subrepresentados en las postulaciones. Por otro lado, los particulares pagados, siendo pocos, son los más efectivos en convertir postulaciones en matrículas.

¿Por qué pasa esto?

La plata es un obstáculo real

---Postular cuesta:

la inscripción, el transporte, los preuniversitarios. Para muchos estudiantes de colegios municipales, estos gastos son una barrera. Además, algunos necesitan trabajar temprano y no 
tienen tiempo para prepararse bien.


---La preparación no es pareja

Los colegios municipales suelen tener menos recursos para preparar a sus estudiantes para las pruebas. Las brechas académicas se van acumulando desde chicos.
Las expectativas también pesan

En algunas familias, ir a la universidad es lo normal y esperado. En otras, no hay nadie cercano que lo haya hecho, y eso influye. Muchos estudiantes se autodescartan porque sienten que "la universidad no es para gente como yo".

---El apoyo institucional varía
La calidad de la orientación vocacional, el acceso a información y las redes de apoyo son muy distintos según el tipo de colegio.

---Las consecuencias

Estamos desperdiciando talento. Hay estudiantes capaces en colegios municipales que ni siquiera postulan. Esto perpetúa las desigualdades de generación en generación y empobrece la diversidad en las universidades. En conclusión:

---La distribución de postulantes es desigual:

los subvencionados lideran en cantidad, los municipales van atrás, y los particulares pagados, aunque son menos, casi todos entran.

Lo grave es que los colegios municipales, que educan a la mayoría, están subrepresentados. Los costos son una barrera, la preparación es desigual, muchos se autodescartan, y el apoyo varía según el colegio.


---Resultado:

desperdiciamos talento y perpetuamos desigualdades. Los números amplifican las brechas existentes en lugar de reflejar el potencial real del país.
Se necesitan medidas concretas: acompañamiento, eliminar costos de postulación, mejor orientación y cupos que consideren el origen. Pero principalmente, hay que atacar el problema desde la educación básica, porque no basta con ayudar a que más postulen.
