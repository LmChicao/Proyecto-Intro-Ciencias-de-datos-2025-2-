# Este archivo esta para descomprimir todo los archivos .zip que contienen los datos unidos por año

Este codigo creara una carpeta para cada año (2023, 2024 y 2025) y una carpeta especifica, ademas los guardara en la carpeta de resultado proyecto, ya que se espera que se usen ppara los jupyter de Resultado 202x.ipynb

Disclaimer : No contiene codigo para descomprimir los datos que se utilizaron para el Analisis Exploratorio

```python
import zipfile
import os


base_path = os.path.dirname(os.getcwd()) 

file_path1 = os.path.join(base_path, 'Resultado proyecto','Union Archivos por Año', 'Año2023.zip')
file_path2 = os.path.join(base_path, 'Resultado proyecto','Union Archivos por Año', 'Año2024.zip')
file_path3 = os.path.join(base_path, 'Resultado proyecto','Union Archivos por Año', 'Año2025.zip')


with zipfile.ZipFile(file_path1, "r") as zip_ref:
    # Extraer todo el contenido
    zip_ref.extractall("Datos_2023")

    # Listar los archivos dentro del ZIP
    print(zip_ref.namelist())
    
with zipfile.ZipFile(file_path2, "r") as zip_ref:
    # Extraer todo el contenido
    zip_ref.extractall("Datos_2024")

    # Listar los archivos dentro del ZIP
    print(zip_ref.namelist())
    
with zipfile.ZipFile(file_path3, "r") as zip_ref:
    # Extraer todo el contenido
    zip_ref.extractall("Datos_2025")

    # Listar los archivos dentro del ZIP
    print(zip_ref.namelist())
```
