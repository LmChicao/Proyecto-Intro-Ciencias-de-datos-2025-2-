# *1) Resumen Proyecto*

Este proyecto de análisis de datos que tiene como objetivo principal investigar las complejas relaciones entre los puntajes académicos y múltiples variables socioeconómicas e institucionales, específicamente analizando cómo se ven reflejados los resultados educativos en relación con los niveles salariales familiares, el ingreso per cápita de los hogares y el tipo de administración escolar (público, privado o subvencionado).

Mediante técnicas de análisis exploratorio y visualización de datos, buscamos identificar patrones, correlaciones y posibles disparidades dentro del sistema educativo, con el propósito de generar insights valiosos que contribuyan a comprender mejor cómo estos factores estructurales influyen en el desempeño académico y, en última instancia, aportar evidencia empírica al debate sobre equidad y calidad educativa.

# *3) Resumen de los resultados*

Los datos revelan que el tipo de colegio donde estudian los jóvenes determina significativamente sus oportunidades de acceso a la educación superior. Los estudiantes de colegios particulares pagados obtienen consistentemente mejores resultados y tienen entre 3 y 4 veces más probabilidades de ingresar a universidades tradicionales que sus pares de colegios municipales.

Esta brecha responde a múltiples factores: los costos asociados a la postulación representan barreras económicas para muchos estudiantes, la preparación académica es desigual desde la educación básica, y existen diferencias en el apoyo familiar y orientación vocacional.

Lo más preocupante es que el sistema, en lugar de nivelar oportunidades, tiende a reproducir las ventajas heredadas. Los colegios municipales, que educan a la mayor parte de la población, aparecen subrepresentados en las postulaciones universitarias, mientras los particulares pagados, aunque numéricamente menores, logran acceder casi en su totalidad.

Estos hallazgos confirman que estamos frente a un desafío estructural que requiere soluciones integrales. No se trata solo de ayudar a que más estudiantes postulen, sino de construir un sistema educativo que realmente valore y desarrolle el potencial de todos los jóvenes, independientemente de su origen.


# *4) ¿Que podria salir mal?*

Nuestro análisis se enfrentó a limitaciones prácticas importantes que es necesario reconocer, ya que no pudimos realizar un seguimiento individual de los estudiantes a lo largo de todo el proceso debido a que los datos de inscripción, rendición de pruebas y postulación se alojaban en sistemas separados que no logramos integrar completamente, a lo que se sumaron formatos inconsistentes entre los diferentes años, complicando los análisis comparativos. 

Es crucial entender que el estudio presenta puntos ciegos significativos, pues solo pudimos analizar a los estudiantes que completaron todo el proceso de postulación, desconociendo lo sucedido con quienes abandonaron en etapas tempranas, y las conclusiones sobre el nivel socioeconómico se basan en inferencias indirectas al carecer de información demográfica explícita. 

Estas limitaciones implican que probablemente estemos subestimando la magnitud real de las desigualdades, ya que los estudiantes en situación de mayor vulnerabilidad posiblemente ni siquiera figuren en nuestros datos; por ello, lo que observamos son síntomas evidentes del problema, pero no capturamos su complejidad completa, representando nuestras conclusiones una versión conservadora de la realidad. 

Para futuras investigaciones, sería fundamental conectar todos los datos del proceso educativo para seguir la trayectoria completa de cada estudiante e incorporar información socioeconómica directa. 

En esencia, nuestro estudio revela una realidad alarmante, pero debemos reconocer que posiblemente estamos viendo solo una parte del problema, por lo que las desigualdades educativas reales probablemente sean aún más profundas de lo que nuestros números logran capturar.
