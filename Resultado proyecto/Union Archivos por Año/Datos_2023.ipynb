```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os
```

# Cargar los datos de 2023


```python
# Inscripcion 2023
base_path = os.path.dirname(os.getcwd()) 
base_path = os.path.dirname(base_path) 
base_path
file_path1 = os.path.join(base_path, 'DEMRE_bdd', '2023','Incripcion','Inscritos_Admisión2023','ArchivoB_Adm2023.csv')
df_2023_inscripcion = pd.read_csv(file_path1, encoding="utf-8", sep=';')

file_path2 = os.path.join(base_path, 'DEMRE_bdd', '2023','Matricula','Matricula_Admisión2023','ArchivoMatr_Adm2023.csv')
df_2023_matricula = pd.read_csv(file_path2, encoding="utf-8", sep=';')

file_path3 = os.path.join(base_path, 'DEMRE_bdd', '2023','Postulacion','PostulaciónySelección_Admisión2023','ArchivoD_Adm2023.csv')
df_2023_postulacion = pd.read_csv(file_path3, encoding="utf-8", sep=';')

file_path4 = os.path.join(base_path, 'DEMRE_bdd', '2023','Rendicion','Rinden_Admisión2023','ArchivoC_Adm2023.csv')
df_2023_Rendicion = pd.read_csv(file_path4, encoding="utf-8", sep=';')    

```
