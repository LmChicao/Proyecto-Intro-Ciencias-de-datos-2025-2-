# Cargar los datos de 2024


```python
base_path = os.path.dirname(os.getcwd()) 
base_path = os.path.dirname(base_path) 
base_path
file_path1 = os.path.join(base_path, 'DEMRE_bdd', '2024','Incripcion','Carpeta 1','Inscritos_Admisión2024','ArchivoB_Adm2024.csv')
df_2024_inscripcion = pd.read_csv(file_path1, encoding="utf-8", sep=';')

file_path2 = os.path.join(base_path, 'DEMRE_bdd', '2024','Matricula','ArchivoMatr_Adm2024.csv')
df_2024_matricula = pd.read_csv(file_path2, encoding="utf-8", sep=';')

file_path3 = os.path.join(base_path, 'DEMRE_bdd', '2024','Postulacion','PostulaciónySelección_Admisión2024','ArchivoD_Adm2024.csv')
df_2024_postulacion = pd.read_csv(file_path3, encoding="utf-8", sep=';')

file_path4 = os.path.join(base_path, 'DEMRE_bdd', '2024','Rendicion','Rinden_Admisión2024','ArchivoC_Adm2024.csv')
df_2024_Rendicion = pd.read_csv(file_path4, encoding="utf-8", sep=';')    
```
